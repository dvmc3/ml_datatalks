## Multidimensional arrays

```python
np.zeros((5,2))
```

```python
list_1 = [1, 2, 3]
list_2 = [4, 5, 6]
list_3 = [7, 8, 9]
n = np.array([list_1, list_2, list_3])
n
```

```python
n[0, 2]
```

```python
n[0, 2] = 10
n
```

```python
n[1]
# passing one index --> getting entire row
```

```python
n[1] = 1
# changing for one index --> changing entire row
n
```

```python
n[:, 1]
# second column
```

```python
n[:, 1] = 0
n
```

## Element-wise operations

```python
a = np.arange(5)
a
```

```python
# Element-wise addition
a = a + 1
a
```

```python
# Element-wise multiplication
a = a * 2
a
```

```python
# Combinations
b = (10 + (a * 2)) ** 2 / 100
b
```

```python
# Element-wise addition of two arrays
c = a + b
c
```
