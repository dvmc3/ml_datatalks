## Q7: Sum of weights

```python
df['origin'].unique()
```

```python
# Numpy matrix X
X = df[df['origin']=='Asia'][:7][['vehicle_weight','model_year']].to_numpy()
```

```python
# View matrix
X
```

```python
# Test X transpose
X.T
```

```python
# Calculation of X transpose * X
XTX = (X.T).dot(X)
```

```python
# Verify that XTX is square
XTX
```

```python
# Inverse matrix
XTX_inv = np.linalg.inv(XTX)
```

```python
# Verify that XTX_inv is indeed the inverse of XTX (product is I)
XTX_inv.dot(XTX)
```

```python
# Create array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
# Calculate w
w = (XTX_inv.dot(X.T)).dot(y)
```

```python
w
```

```python
# Solution
w.sum()
```
